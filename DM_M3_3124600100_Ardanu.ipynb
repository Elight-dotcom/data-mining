## Normalisasi

### Min-Max

```python
import pandas as pd
import numpy as np

def min_max_norm(data):
    return (data - np.min(data)) / (np.max(data) - np.min(data))

data_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/Salinan titanic.csv'
dataset = pd.read_csv(data_path)

dataset['Age'] = dataset['Age'].fillna(dataset.groupby('Pclass')['Age'].transform('mean'))

num_cols = ['Age', 'Fare', 'SibSp', 'Parch']

for col in num_cols:
  dataset[col] = min_max_norm(dataset[col])

print(dataset[num_cols].head())
print(dataset[num_cols].describe())
```

### Z-Score

```python
import pandas as pd
import numpy as np

def z_score_norm(data):
    return (data - np.mean(data)) / np.std(data)

data_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/Salinan titanic.csv'
dataset = pd.read_csv(data_path)

dataset['Age'] = dataset['Age'].fillna(dataset.groupby('Pclass')['Age'].transform('mean'))

num_cols = ['Age', 'Fare', 'SibSp', 'Parch']

for col in num_cols:
  dataset[col] = z_score_norm(dataset[col])

print(dataset[num_cols].head())
print(dataset[num_cols].describe())
```

### Sigmoid

```python
import pandas as pd
import numpy as np

def sigmoid_norm(data):
    z = (data - np.mean(data)) / np.std(data)
    return (1 - np.exp(-z)) / (1 + np.exp(-z))

data_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/Salinan titanic.csv'
dataset = pd.read_csv(data_path)

dataset['Age'] = dataset['Age'].fillna(dataset.groupby('Pclass')['Age'].transform('mean'))

num_cols = ['Age', 'Fare', 'SibSp', 'Parch']

for col in num_cols:
  dataset[col] = sigmoid_norm(dataset[col])

print(dataset[num_cols].head())
print(dataset[num_cols].describe())
```
