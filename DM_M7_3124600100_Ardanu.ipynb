## Menampilkan dataset heart.csv

### Menambahkan header

```python
import pandas as pd
import numpy as np

data_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/heart.csv'

h = [
    "Age",
    "sex",
    "chest pain type",
    "resting blood pressure",
    "serum cholestoral in mg/dl",
    "fasting blood sugar > 120 mg/dl",
    "resting electrocardiographic results",
    "maximum heart rate achieved",
    "exercise induced angina",
    "oldpeak = ST depression induced by exercise relative to rest",
    "the slope of the peak exercise ST segment",
    "number of major vessels (0-3) colored by flourosopy",
    "thal",
    "heart_disease"
]
dataset = pd.read_csv(data_path, header=None, names=h)

dataset
```

```python
# Cek jumlah kategori di kolom katgoerikal
sex_count = dataset['sex'].value_counts()
chest_pain_type_count = dataset['chest pain type'].value_counts()
resting_electrocardiographic_results_count = dataset['resting electrocardiographic results'].value_counts()
exercise_induced_angina_count = dataset['exercise induced angina'].value_counts()
the_slope_of_the_peak_exercise_ST_segment_count = dataset['the slope of the peak exercise ST segment'].value_counts()
thal_count = dataset['thal'].value_counts()

print("Jumlah kategori kolom sex:")
print(sex_count)
print("\nJumlah kategori kolom chest pain type:")
print(chest_pain_type_count)
print("\nJumlah kategori kolom resting electrocardiographic results:")
print(resting_electrocardiographic_results_count)
print("\nJumlah kategori kolom exercise induced angina:")
print(exercise_induced_angina_count)
print("\nJumlah kategori kolom the slope of the peak exercise ST segment:")
print(the_slope_of_the_peak_exercise_ST_segment_count)
print("\nJumlah kategori kolom thal:")
print(thal_count)
```

```python
# Cek missing value
print(dataset.isna().sum())

# Cek duplicate
print("\nData duplikat:", dataset.duplicated().sum())
```

### Pemisahan Fitur

```python
features = dataset.copy().drop(columns=['heart_disease'])
target = dataset['heart_disease']

target
```

### Encoding

#### Encoding Features

```python
# Encoding biary
features['sex'] = features['sex'].map({'female': 0, 'male': 1})
features['exercise induced angina'] = features['exercise induced angina'].map({'no': 0, 'yes': 1})

# Encoding ordinal
features['the slope of the peak exercise ST segment'] = features['the slope of the peak exercise ST segment'].map({
    'upsloping': 0,
    'flat': 1,
    'downsloping': 2
})

# OHE
nominal_column = ['chest pain type', 'resting electrocardiographic results', 'thal']
features = pd.get_dummies(features, columns=nominal_column, drop_first=True)

features
```

#### Encoding Target

```python
target = target.map({'Absence': 0, 'Presence': 1})

target
```

## Inisialisasi Normalisasi, Klasifikasi Model dan Validation Model

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, FunctionTransformer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold, LeaveOneOut, cross_val_score

# Dictionary Normalisasi
normalization = {
    'Min-Max': MinMaxScaler(),
    'Z-Score': StandardScaler(),
    'Sigmoidal': FunctionTransformer(lambda x: 1 / (1 + np.exp(-x)))
}

# Dictionary Model
models = {
    'k-NN (k=1)': KNeighborsClassifier(n_neighbors=1),
    'k-NN (k=5)': KNeighborsClassifier(n_neighbors=5),
    'k-NN (k=9)': KNeighborsClassifier(n_neighbors=9),
    'Decision Tree': DecisionTreeClassifier(random_state=42)
}

# Dictionary Validasi
cv_methods = {
    'k-Fold (k=4)': StratifiedKFold(n_splits=4, shuffle=True, random_state=42),
    'k-Fold (k=10)': StratifiedKFold(n_splits=10, shuffle=True, random_state=42),
    'LOO': LeaveOneOut()
}

# List kosong untuk hasil
result = []
```

## Pengerjaan

```python
from sklearn.pipeline import Pipeline

for cv_name, cv in cv_methods.items():
    for scaler_name, scaler in normalization.items():
        for model_name, model in models.items():

            # Gabung scaler dan model dalam satu Pipeline
            pipeline = Pipeline([
                ('scaler', scaler),
                ('model', model)
            ])

            # Jalankan eksperimen dengan cross_val_score
            akurasi_fold = cross_val_score(pipeline, features, target, cv=cv, scoring='accuracy')

            # Hitung rata-rata akurasi dan error ratio
            rata_akurasi = akurasi_fold.mean()
            error_ratio = 1 - rata_akurasi

            # Simpan ke dalam list
            result.append({
                'Validasi': cv_name,
                'Normalisasi': scaler_name,
                'Model': model_name,
                'Error Ratio': error_ratio
            })
```

```python
# Urutkan berdasarkan error ratio paling kecil
df_result = pd.DataFrame(result)
df_hasil_urut = df_result.sort_values(by='Error Ratio').reset_index(drop=True)

df_hasil_urut
```
