## Import Library

```python
import pandas as pd
```

## Menampilkan dataset titanic.csv

```python
data_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/titanic.csv'
dataset = pd.read_csv(data_path)

print(dataset)
```

## Menampilkan dataset titanic_test.csv

```python
data_test_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/titanic_test.csv'

test_dataset = pd.read_csv(data_test_path)

print(test_dataset)
```

## Menentukan data train dari fitur (Age, Fare) yang sudah dihilangkan missing valuenya

```python
subset_train = dataset[['Age', 'Fare']]

pos_missing_train = subset_train[subset_train.isnull().any(axis=1)].index.tolist()
train_data = subset_train.dropna()

print("Posisi data train yang hilang:", pos_missing_train)
print(train_data)
```

## Menentukan data test dari fitur (Age, Fare) yang sudah dihilangkan missing valuenya

```python
subset_test = test_dataset[['Age', 'Fare']]

pos_missing_test = subset_test[subset_test.isnull().any(axis=1)].index.tolist()
test_data = subset_test.dropna()

print("Posisi data test yang hilang:", pos_missing_test)
print(test_data)
```

## Ambil dataset kolom kelas (Survived), yang bukan pos_missing_train

```python
train_label = dataset['Survived'].drop(index=pos_missing_train)

print("\nIsi train_label:")
print(train_label)
```

## Ambil dataset kolom kelas (Survived), yang bukan pos_missing_test

```python
data_test_label_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/titanic_testlabel.csv'

data_test_label = pd.read_csv(data_test_label_path)
test_label = data_test_label['Survived'].drop(index=pos_missing_test)

print("\nIsi test_label:")
print(test_label)
```

## Lakukan normalisasi pada train_data dengan Min-Max 0-1 (catat nilai min dan max setiap atribut)

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

train_data = pd.DataFrame(scaler.fit_transform(train_data),
                               columns=train_data.columns,
                               index=train_data.index)

print("\nHasil train_data setelah normalisasi 0-1:")
print(train_data)
```

```python
test_data = pd.DataFrame(scaler.transform(test_data),
                              columns=test_data.columns,
                              index=test_data.index)

print("Hasil test_data setelah normalisasi 0-1:")
print(test_data)
```

```python
from sklearn.neighbors import KNeighborsClassifier

class_result = pd.DataFrame(index=test_data.index)

for k in range(1, 16):
    knn = KNeighborsClassifier(n_neighbors=k)

    knn.fit(train_data, train_label)

    predict = knn.predict(test_data)

    class_result[f'k={k}'] = predict

print("Hasil klasifikasi test_data untuk k=1 sampai 15:")
print(class_result)
```

```python
error_ratios = {}

total_data = len(test_label)

for k in range(1, 16):
    kolom_k = f'k={k}'

    jumlah_error = (class_result[kolom_k].values != test_label.values).sum()

    error_ratio = jumlah_error / total_data

    error_ratios[kolom_k] = error_ratio

df_error = pd.DataFrame(list(error_ratios.items()), columns=['Nilai K', 'Error Ratio'])

print("Error Ratio untuk masing-masing k:")
print(df_error)
```
