### Mengambil dataset

```python
import pandas as pd
import numpy as np

data_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/titanic.csv'

dataset = pd.read_csv(data_path)
dataset['Sex'] = dataset['Sex'].map({'female': 0, 'male': 1})

dataset
```

```python
data = dataset[['Sex', 'Age', 'Pclass', 'Fare', 'Survived']]
data
```

```python
train_data = dataset.loc[dataset['Age'].notnull(), ['Sex', 'Pclass', 'Fare', 'Survived']]
train_data
```

```python
train_label = dataset.loc[dataset['Age'].notnull(), 'Age']
train_label
```

```python
test_data = dataset.loc[dataset['Age'].isnull(), ['Sex', 'Pclass', 'Fare', 'Survived']]
test_data
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
train_data = pd.DataFrame(scaler.fit_transform(train_data), columns=train_data.columns)

print('Nilai min: \n', train_data.min(axis=0))
print('Nilai max: \n', train_data.max(axis=0))
```

```python
test_data_scaled = pd.DataFrame(scaler.transform(test_data), columns=test_data.columns)

print(test_data[:10])
```

```python
from sklearn.neighbors import KNeighborsRegressor

knn = KNeighborsRegressor(n_neighbors=3)
knn.fit(train_data, train_label)

class_result = knn.predict(test_data)
print(class_result)
```

```python
data.loc[dataset['Age'].isnull(), 'Age'] = class_result

print("Jumlah data kosong pada kolom Age sekarang:", data['Age'].isnull().sum())
```

```python
test_data_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/titanic_test.csv'

test_dataset = pd.read_csv(test_data_path)
test_dataset['Sex'] = test_dataset['Sex'].map({'female': 0, 'male': 1})
test_dataset
```

```python
train_data = data[['Sex', 'Age', 'Pclass', 'Fare']]

train_data
```

```python
train_label = data['Survived']

train_label
```

```python
test_data = test_dataset[['Sex', 'Age', 'Pclass', 'Fare']].dropna()

print('Jumlah data null', test_data.isnull().sum().sum())
```

```python
test_label_path = '/content/drive/MyDrive/PENS/Semester 5/Data Mining/titanic_testlabel.csv'

test_label = pd.read_csv(test_label_path)
test_label = test_label.loc[test_data.index]
test_label = test_label['Survived']

test_label
```

```python
train_data = scaler.fit_transform(train_data)

print('Nilai min train_data', train_data.min(axis=0))
print('Nilai max train_data', train_data.max(axis=0))
```

```python
test_data = scaler.transform(test_data)

test_data
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(train_data, train_label)

class_result = knn.predict(test_data)
print(class_result)
```

```python
total_data = test_label.shape[0]
error = (test_label.values.ravel() != class_result).sum()

print(f"Jumlah Error: {error} dari {total_data} data")
```

```python
error_ratio = error / total_data

print(f"Error Ratio: {error_ratio * 100:.2f}%")
```
